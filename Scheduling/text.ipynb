```python
def static_schedule(jobs , workers):
    n = len(jobs)
    w = workers
    
    job = []
    for i in range(w):
        a = []
        for j in range(n//w):
            a.append(jobs[j + i * (n // w)])
        job.append(a)    

    return job
print(static_schedule([3, 8, 2, 6, 9, 1, 5, 7, 4, 10], 4))
```
