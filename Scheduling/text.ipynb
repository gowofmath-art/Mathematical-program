```python
def static_schedule(jobs , workers):
    n = len(jobs)
    w = workers
    
    job = []
    for i in range(w):
        a = []
        for j in range(n//w):
            a.append(jobs[j + i * (n // w)])
        job.append(a)    

    return job
print(static_schedule([3, 8, 2, 6, 9, 1, 5, 7, 4, 10], 4))
```

```python
def static_schedule(jobs , workers):
    n = len(jobs)
    w = workers

    job = []
    start = 0

    for i in range(w):
        a = []
        x = n//w
        if i < n % w:
            x = x +1

        for j in range(x):
            a.append(jobs[start + j])

        job.append(a)
        start = start + x

    return job

print(static_schedule([3, 8, 2, 6, 9, 1, 5, 7, 4, 10], 4))
```
