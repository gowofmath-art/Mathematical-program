提示：找出那年那個月1日是星期幾

由於一開始我沒頭緒，覺得有點困難，因此我先簡化問題，從單一特例下手，再慢慢推廣至原本的難題。  
具體的規劃任務如下：

2. 輸出今年2026年4月的月曆：換其他月，1號不在星期日的情況  
3. 完成今年2026年所有月的月曆：必須輸入月，即可輸出2026該月的月曆
4. 完成其他年份：考慮閏年的問題

6. 尋找最佳解：使用蔡勒公式（Zeller's congruence）

```python
# 3.完成今年2026年所有月的月曆
# 必須輸入月，即可輸出2026該月的月曆
# 2026/1/1是星期四
# 其想法就是真的去算從2026/1/1開始，到指定月之前到底過了幾天，才可以決定start_day在星期幾


m = int(input())
print(f'這是2026年{m}月月曆')

D = '日一二三四五六'
for n in D:
    print(f'{n:>2}', end="")
print()

def calendar(m):
    mon = [31,28,31,30,31,30,31,31,30,31,30,31]
    f = 4
    day = sum((mon[i]) for i in range(m-1) )
    start_day = (f + day)%7
    print("   "*start_day, end="")
    for i in range(1, (mon[m-1])+1):
        print(f'{i:>2}', end=" ")
        if (i+start_day) % 7 == 0:
            print()

calendar(m)
```

```python
# 4.完成其他年份：考慮閏年的問題 
# 每四年一閏、百年不閏、四百年又閏

y = int(input())
m = int(input())
print(f'這是{y}年{m}月的月曆')

D = '日一二三四五六'
for n in D:
    print(f'{n:>2}', end="")
print()

def calendar(y,m):
    mon = [31,28,31,30,31,30,31,31,30,31,30,31]
    f = 1
    if (y%4 == 0 and y%100 != 0 or y%400 ==0) :
        if m <=2 :
            day = 365*(y-1)+(y//4-y//100+y//400)+sum((mon[i]) for i in range(m-1) )-1
            start_day = (f + day)%7
            print("   "*start_day, end="")
            if m==1:
                for i in range(1, (mon[m-1])+1):
                    print(f'{i:>2}', end=" ")
                    if (i+start_day) % 7 == 0:
                        print()
            else:
                for i in range(1, (mon[m-1])+1+1):
                    print(f'{i:>2}', end=" ")
                    if (i+start_day) % 7 == 0:
                        print()
            print()            
            print("閏年")
        else:
            day = 365*(y-1)+(y//4-y//100+y//400)+sum((mon[i]) for i in range(m-1) )
            start_day = (f + day)%7
            print("   "*start_day, end="")
            for i in range(1, (mon[m-1])+1):
                print(f'{i:>2}', end=" ")
                if (i+start_day) % 7 == 0:
                    print()
            print() 
            print("閏年")
    else:
        day = 365*(y-1)+(y//4-y//100+y//400)+sum((mon[i]) for i in range(m-1) )
        start_day = (f + day)%7
        print("   "*start_day, end="")
        for i in range(1, (mon[m-1])+1):
            print(f'{i:>2}', end=" ")
            if (i+start_day) % 7 == 0:
                print()
        print()
        print("非閏年")

calendar(y,m)
```

```python
# 5. 優化版

y = int(input())
m = int(input())
print(f'這是{y}年{m}月的月曆')

D = '日一二三四五六'
for n in D:
    print(f'{n:>2}', end="")
print()

def is_leap(y):
    return (y%4 == 0 and y%100 != 0 or y%400 == 0)
def calendar(y, m):
    mon = [31,28,31,30,31,30,31,31,30,31,30,31]
    if is_leap(y):
        mon[1] = 29
    f = 1  
    day = 365*(y-1) + ((y-1)//4 - (y-1)//100 + (y-1)//400) + sum(mon[:m-1])
    start_day = (f + day) % 7
    print("   " * start_day, end="")
    for i in range(1, mon[m-1] + 1):
        print(f'{i:>2}', end=" ")
        if (i + start_day) % 7 == 0:
            print()
calendar(y,m)
```

```python
# 6.使用蔡勒公式（Zeller's congruence）
# 直接告訴星期幾 (星期日 = 0 依序下去)
# 在輸出  幾年幾月的月曆

y = int(input())
m = int(input())
print(f'這是{y}年{m}月的月曆')

D = '日一二三四五六'
for n in D:
    print(f'{n:>2}', end="")
print()

def Zeller(y,m,d):
    if m < 3:
        m += 12
        y -= 1
    y1 = y % 100
    y2 = y // 100
    w = (y1 + y1//4 + y2//4 - 2*y2 + (13*(m+1)//5) + d - 1) % 7
    return w

def is_leap(y):
    return (y%4 == 0 and y%100 != 0) or (y%400 == 0)

def calendar(y, m):
    mon = [31,28,31,30,31,30,31,31,30,31,30,31]
    if is_leap(y):
        mon[1] = 29
    start_day = Zeller(y, m, 1)
    print("   " * start_day, end="")
    for i in range(1, mon[m-1] + 1):
        print(f'{i:>2}', end=" ")
        if (i + start_day) % 7 == 0:
            print()

calendar(y, m)
```
