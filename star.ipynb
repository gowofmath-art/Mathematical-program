## 題目 7 (菱形) -3 事後回顧：  
過了兩三個月回頭看當初的程式碼，現在可以好好回答。  
此處的 else 因縮排錯誤，實際上屬於 for y，而不是 if。  
Python 的 for...else 會在迴圈正常結束後執行一次，因此每列只多印出一個空格，沒有替不符合條件的格點輸出空白。  

這會讓它從「殘留錯誤」升級成「有教學價值的錯誤案例」。  

```python
n = 9
a = n // 2

for x in range(n):
    for y in range(n):
        if abs(x-a) + abs(y-a) <= a:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

## 題目 9 (正方形邊框) 的事後回顧-為什麼這個條件無法畫出正方形邊框？  

其實題目與我print出來的有出路，這是依照中心點曼哈頓距離篩選後留下的四角區域。  
因為它揭示了：
$∣x−a∣+∣y−a∣$

描述的是菱形距離，不是正方形距離。  
正方形更適合用切比雪夫距離：

$max(∣x−a∣,∣y−a∣)=a$




```python
# 題目 10 (圓形)


n = 30
a = n//2
for x in range(n):
    for y in range(n):
        if (x-a)**2 + (y-a)**2 <= (a)**2:
            print("*", end="")
        else:
            print(" ", end="")
    print()


```
